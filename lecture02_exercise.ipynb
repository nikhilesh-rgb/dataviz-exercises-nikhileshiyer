## Setup — Run this cell first


```python
import pandas as pd
import numpy as np

df = pd.read_csv(r'C:\Users\Dell\world_happiness_2023.csv')

df = df.rename(columns={
    'Country name': 'Country',
    'Regional indicator': 'Region',
    'Ladder score': 'Happiness_Score',
    'Logged GDP per capita': 'GDP',
    'Social support': 'Social_Support',
    'Healthy life expectancy': 'Life_Expectancy',
    'Freedom to make life choices': 'Freedom',
    'Perceptions of corruption': 'Corruption',
})

df = df[['Country', 'Region', 'Happiness_Score', 'GDP', 'Social_Support',
         'Life_Expectancy', 'Freedom', 'Generosity', 'Corruption']]

print(f"Dataset: {len(df)} countries, {len(df.columns)} columns")
df.head()
```

```python
import plotly.express as px
import plotly.graph_objects as go

# Explore the dataset before you start
print("Regions in dataset:")
print(df['Region'].value_counts())
print("\nScore range:", df['Happiness_Score'].min(), "–", df['Happiness_Score'].max())
print("\nBottom 10 countries:")
print(df.nsmallest(10, 'Happiness_Score')[['Country','Region','Happiness_Score']])
```

## Task 1 — Regional Comparison Bar Chart

**What to build:** A horizontal bar chart showing the **average happiness score by region**, sorted from highest to lowest.

**Requirements:**
- Horizontal orientation (category names are long)
- Sorted by score, descending (so the happiest region is at the top)
- Zero baseline on x-axis
- At least one design choice that goes beyond the Plotly default (colour, annotation, labels, etc.)
- An insight title that answers: *which region stands out and why does it matter?*

**Hint:** Use `df.groupby('Region')['Happiness_Score'].mean()` to compute the averages.


```python
# Task 1: Regional comparison bar chart
# -------------------------------------

# Step 1: Compute average happiness score by region
region_avg = (df.groupby('Region')['Happiness_Score']
              .mean()
              .reset_index()
              .sort_values('Happiness_Score'))

print(region_avg)

# Step 2: Build your chart
# YOUR CODE HERE

import matplotlib.pyplot as plt

plt.figure(figsize=(12, 8))

plt.barh(
    region_avg['Region'],
    region_avg['Happiness_Score']
)

# Highest score at the top
plt.gca().invert_yaxis()

# Zero baseline
plt.xlim(left=0)

# Axis labels
plt.xlabel('Average Happiness Score')
plt.ylabel('Region')

# Insight title
plt.title(
    'Western Europe Stands Out as the Happiest Region — Highlighting a Strong Regional Well-Being Advantage',
    fontsize=14,
    fontweight='bold'
)

# Add score labels
for i, score in enumerate(region_avg['Happiness_Score']):
    plt.text(
        score + 0.03,
        i,
        f'{score:.2f}',
        va='center'
    )

plt.tight_layout()
plt.show()
```

## Task 2 — Bottom vs. Top: A Contrast Story

**What to build:** A bar chart that highlights the **gap between the happiest and least happy countries**, focusing on a specific insight.

**Requirements:**
- Show the **top 8 AND bottom 8 countries** together (16 bars total)
- Use **colour** to distinguish the two groups (not Plotly's default rainbow)
- Add a **visual separator or annotation** that emphasises the gap
- Insight title that tells the story of the gap

**Hint:** Use `pd.concat([df.nlargest(8,'Happiness_Score'), df.nsmallest(8,'Happiness_Score')])` to get both groups.

**Stretch goal:** Add a vertical reference line showing the global average.


```python
# Task 2: Top 8 vs. Bottom 8 contrast
# ------------------------------------

# Step 1: Get top and bottom countries
top8 = df.nlargest(8, 'Happiness_Score').copy()
top8['Group'] = 'Top 8'
bottom8 = df.nsmallest(8, 'Happiness_Score').copy()
bottom8['Group'] = 'Bottom 8'

combined = pd.concat([bottom8, top8]).sort_values('Happiness_Score')
global_avg = df['Happiness_Score'].mean()
print(f"Global average: {global_avg:.2f}")

# Step 2: Build your chart
# YOUR CODE HERE

import matplotlib.pyplot as plt

plt.figure(figsize=(14, 9))

# Separate the two groups
bottom_data = combined[combined['Group'] == 'Bottom 8']
top_data = combined[combined['Group'] == 'Top 8']

# Plot bottom 8
plt.barh(
    bottom_data['Country'],
    bottom_data['Happiness_Score'],
    label='Bottom 8'
)

# Plot top 8
plt.barh(
    top_data['Country'],
    top_data['Happiness_Score'],
    label='Top 8'
)

# Global average reference line
plt.axvline(
    global_avg,
    linestyle='--',
    linewidth=2,
    label=f'Global Average ({global_avg:.2f})'
)

# Add score labels
for i, row in enumerate(combined.itertuples()):
    plt.text(
        row.Happiness_Score + 0.05,
        i,
        f'{row.Happiness_Score:.2f}',
        va='center'
    )

# Labels
plt.xlabel('Happiness Score')
plt.ylabel('Country')

# Insight title
plt.title(
    'A Large Happiness Gap Separates the Happiest and Least Happy Countries',
    fontsize=15,
    fontweight='bold'
)

# Legend
plt.legend()

# Zero baseline
plt.xlim(left=0)

plt.tight_layout()
plt.show()

```

**Task 3 (stretch):** Build a **grouped bar chart** comparing 2 sub-factors (e.g. `GDP_per_capita` and `Freedom`) across the 5 most populated regions. Use colour meaningfully and write an insight title.

Regions to include: `'Western Europe'`, `'Latin America'`, `'East Asia'`, `'Sub-Saharan Africa'`, `'South Asia'`


```python
# Stretch goal — grouped bar chart
# YOUR CODE HERE
# Task 3: GDP vs Freedom across selected regions
# ----------------------------------------------

selected_regions = [
    'Western Europe',
    'Latin America',
    'East Asia',
    'Sub-Saharan Africa',
    'South Asia'
]

region_data = df[df['Region'].isin(selected_regions)]

region_avg = (region_data.groupby('Region')[['GDP', 'Freedom']]
              .mean()
              .reindex(selected_regions))

print(region_avg)

# Build grouped bar chart
x = np.arange(len(region_avg))
width = 0.35

plt.figure(figsize=(12, 7))

plt.bar(
    x - width/2,
    region_avg['GDP'],
    width,
    label='GDP per Capita'
)

plt.bar(
    x + width/2,
    region_avg['Freedom'],
    width,
    label='Freedom'
)

# Labels
plt.xlabel('Region')
plt.ylabel('Average Score')
plt.xticks(x, region_avg.index, rotation=20)

# Insight title
plt.title(
    'Western Europe Leads in GDP and Freedom, Highlighting Stronger Socioeconomic Conditions',
    fontsize=14,
    fontweight='bold'
)

plt.legend()

plt.tight_layout()
plt.show()
```
